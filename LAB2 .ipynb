```python
password=input("Enter Your Password:")
if len(password)< 8:
    print("Weak: Min 8 characters")
if not any (not c.isalnum() for c in password):
    print("Weak:Please add a Special character for ex.(@ $ % & * )")
if not any (c.isdigit() for c in password):
    print("Weak:Please Enter a Digit")
    
if (len(password)>=8
and any (c.isalnum() for c in password)
and any (not c.isdigit() for c in password)):
    print("Password Strength: Strong")
    
```

```python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

for celsius in range(0, 101):
    fahrenheit = celsius_to_fahrenheit(celsius)
    print(celsius, "°C =", fahrenheit, "°F")
```

```python
def factorial(number):
    if number < 0 or not isinstance(number, int):
        print("Error: Please enter a non-negative integer.")
        return
    result = 1
    for i in range(1, number + 1):
        result = result * i
    return result
number = int(input("Enter a non-negative integer: "))
answer = factorial(number)
if answer is not None:
    print("Factorial:", answer)
```

```python
def is_prime(number):
    if number < 2:
        return False
    for i in range(2, number):
        if number % i == 0:
            return False
    return True
number = int(input("Enter a number: "))
if is_prime(number):
    print("Prime")
else:
    print("Not prime")
```
